```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t_step = 50
N = 10
pstate = np.zeros(N)

for i in range(t_step):
    r = np.random.randint(2)
#    print(r)
    x = np.where(pstate == r)[0]
#    print(x)

    if np.size(x) != 0:
        rnd = np.random.choice(x)
        pstate[rnd] = 1 - pstate[rnd]

#    print(pstate)

# Holy crap! This works!
# So, the idea is, I randomly generate a 0 or a 1
# The tricky part was adding the [0] at the end of the np.where function, if not the program doesn't work
# https://stackoverflow.com/questions/48686381/how-to-get-a-list-of-indexes-selected-by-a-specific-value-efficiently-with-numpy
# And then basically the loop checks if the array I selected is not empty... if it is empty, by default it does nothing!
# And if it is not empty, it randomly selects an entry and flips it!
# I haven't been able to get the left side box empty but I am assuming it will work!
```
