```python
# add to the other notebook
# this generates random dice throws from 1 to 6 and then check how many times left is larger than right...
# this is 10 times faster than not using arrays

N = 100
r = np.random.randint(1, 6, size=(2, N))
left = r[0, :]
right = r[1, :]
#print(left)
#print(right)

success = left > right
#print(success)
M = np.sum(success)
probability = M / N
print(probability)

# the theoretical answer is found:
combinations = [(black, green)
                for black in range(1, 7)
                for green in range(1, 7)]
print(combinations)

success = [black > green for black, green in combinations]
#print(success)
print(np.size(success))
M = sum(success)
print(M / np.size(success))
```
