# Some basics

```python
from random import randrange

dice_1 = randrange(6)
dice_2 = randrange(6)
print("Dice 1 gives:", dice_1)
print("Dice 2 gives:", dice_2)

# A dice cannot give 0 though, so this is not great...
```

```python
N = 1000000
x = 0
y = 0

for i in range(N):
    dice_1 = randrange(1, 7)       # The first value is included, the second one isn't.
    dice_2 = randrange(1, 7)       # Clearly randrange produces integers.
    
    if dice_1 == dice_2:
        x += 1
        if dice_2 == 6:            # I like this bit because it already knows the two dice #s are equal
            y += 1                 # So it counts every time they're both 6

print("This is the percentage of times we get the same number:", x/1000000)
print("This is the percentage of times we get the number 6 twice:", y/1000000)

# note - use xrange(N) instead of range(N) in the for loop if you don't want to store N numbers in memory! 
# (for large N)
# Is 2.8 percent the right probability? Need to check this!
```

```python
N = 1000000
x = 0
y = 0

for i in range(N):
    dice_1 = randrange(7)
    dice_2 = randrange(7)
    
    if dice_1 == dice_2 == 6:
        x += 1

print("This is the percentage of times we get the number 6 twice:", x/1000000)
```

```python
# add to the other notebook
# this generates random dice throws from 1 to 6 and then check how many times left is larger than right...
# this is 10 times faster than not using arrays

N = 100
r = np.random.randint(1, 6, size=(2, N))
left = r[0, :]
right = r[1, :]
#print(left)
#print(right)

success = left > right
#print(success)
M = np.sum(success)
probability = M / N
print(probability)

# the theoretical answer is found:
combinations = [(black, green)
                for black in range(1, 7)
                for green in range(1, 7)]
print(combinations)

success = [black > green for black, green in combinations]
#print(success)
print(np.size(success))
M = sum(success)
print(M / np.size(success))
```
