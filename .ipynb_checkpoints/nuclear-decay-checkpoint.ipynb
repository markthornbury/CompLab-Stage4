```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
n_thallium = 1000
n_lead = 0
tau = 3.053 * 60
h = 1.0
p = 1 - 2**(-h/tau)
tmax = 1000

# Plot points
t_points = np.arange(0, tmax, h)
thallium_points = []
lead_points = []

# Main loop
for i in t_points:
    thallium_points.append(n_thallium)
    lead_points.append(n_lead)

    # Decay calculation
    decay = 0
    for j in range(n_thallium):
        if random.random() < p:
            decay += 1
    n_thallium -= decay
    n_lead += decay

#Graph
plt.plot(t_points, thallium_points)
plt.plot(t_points, lead_points)
plt.grid(True)
```
