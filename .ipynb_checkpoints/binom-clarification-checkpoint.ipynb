```python
import numpy as np
#import scipy.special
import scipy.stats
import matplotlib.pyplot as plt
```

```python
a = np.random.binomial(1, 0.5, size=10)
b = np.random.binomial(10, 0.5)
c = np.random.binomial(10, 0.5, size=10)
print(a)
print(b)
print(c)
plt.hist(a)
plt.hist(c)
```

```python
# I could also generate individual flips for 100 experiments, each containing 10 flips.
# And then count how many times there is a 1 in each row.

flips = np.random.binomial(1, 0.5, size=(100, 10))
success = np.count_nonzero(flips, axis = 1)
print(success)
```
